# Día 2 · El mismo panel con Plotly, Streamlit y Dash

En el [notebook 16](16_panel_siniestros_dia_2.ipynb) armamos un panel con subplots y exportamos `resumen_siniestros_panel.csv`. Aquí lo reutilizamos sin repetir la limpieza. Veremos tres maneras de mostrar el mismo indicador, la línea por mes, las barras por zona y el mapa de calor.

1. **Plotly en el notebook.** Una figura con varios subplots. Podemos acercarnos y leer los valores al pasar el ratón.
2. **Streamlit.** Los filtros de la barra lateral vuelven a ejecutar el script y actualizan la figura.
3. **Dash.** Un callback conecta los filtros con la figura. El resto de la página permanece en su sitio.

## 1 · Preparar el entorno

Ejecuta la exportación del 16 antes de seguir. Desde Jupyter o VS Code puedes abrir este notebook en el repositorio. En Colab sube al entorno el CSV exportado y los archivos `panel_siniestros.py` y `app_dash.py`. Para el panel en el notebook, instala `%pip install pandas plotly` si faltan. Instala `dash>=2.18` si vas a probar Dash.

Streamlit se muestra con un servidor local desde la terminal de tu ordenador. Para una clase que solo usa Colab, puedes mostrar la figura y la versión Dash en el notebook.

````python
from pathlib import Path
import sys

inicio = Path.cwd()
opciones = [inicio / 'apps', inicio.parent / 'apps', inicio / 'PySpark' / 'apps', inicio]
carpeta_apps = next((p for p in opciones if (p / 'panel_siniestros.py').is_file()), None)
if carpeta_apps is None:
    raise FileNotFoundError('Abre el notebook en el repositorio o sube panel_siniestros.py a Colab.')
sys.path.insert(0, str(carpeta_apps))

from panel_siniestros import cargar_datos, figura_panel, resumir

datos = cargar_datos()
print(f"Total de expedientes del CSV: {datos['expedientes'].sum():,}")
datos.head()
````

## 2 · Plotly dentro del notebook

La función `figura_panel` filtra una vez y crea todos los subplots con el mismo resumen. Al igual que en el 16, si elegimos un distrito las barras muestran sus barrios. La selección del ejemplo debe devolver **1.099 expedientes**.

````python
figura_panel(datos, distrito='Eixample', periodo=(1, 6)).show()
````

## 3 · Streamlit desde la terminal

Abre una terminal en la carpeta **PySpark** del repositorio. Ejecuta:

```text
python -m pip install pandas plotly streamlit
python -m streamlit run apps/app_streamlit.py --server.address 127.0.0.1
```

Abre la dirección local que indique la terminal. Cambia el distrito y mueve los dos extremos del intervalo de meses. Streamlit lee los filtros, vuelve a ejecutar el script y muestra la figura actualizada. Para detenerlo, pulsa **Ctrl+C** en la terminal.

En [app_streamlit.py](../apps/app_streamlit.py), busca tres líneas: `selectbox` elige el distrito, `slider` elige los meses y `plotly_chart` dibuja el resultado.

## 4 · Dash en el notebook o desde la terminal

Con Dash la pantalla se define en `app.layout`. El callback dice qué datos entran y qué parte de la pantalla debe actualizarse. En [app_dash.py](../apps/app_dash.py), los dos filtros son entradas y la figura es la salida.

Para mostrar Dash en el notebook, instala `%pip install "dash>=2.18"` y cambia `mostrar_dash` a `True` en la celda siguiente. Si tu entorno no admite el panel incrustado, usa la terminal local:

```text
python -m pip install pandas plotly "dash>=2.18"
python apps/app_dash.py
```

````python
mostrar_dash = False
if mostrar_dash:
    from app_dash import crear_app
    crear_app(datos).run(jupyter_mode='inline', jupyter_height=950, debug=False, port=8051)
````

Selecciona **Eixample, enero a junio** en cada versión. El indicador debe mostrar **1.099**. Comprueba que la línea solo tiene seis meses y que las barras muestran barrios. Después cambia a **todos los distritos, todo el año**. Deben aparecer **7.721 expedientes**.

Usa la figura en el notebook para explorar y explicar. Streamlit permite construir una interfaz con poco código. Dash permite definir qué componente se actualiza con cada interacción. Las tres vistas usan el [mismo resumen y los mismos subplots](../apps/panel_siniestros.py).

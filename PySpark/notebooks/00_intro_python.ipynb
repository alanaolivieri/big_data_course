# Primeros pasos con Python

Vamos a calcular la duración de varias sesiones de un curso y a organizar los datos en una tabla. Ejecuta las celdas de arriba abajo para que las variables estén disponibles cuando las necesites.

## 1. Mostrar resultados con `print()`

`print()` muestra valores en la salida de la celda. Una *f-string* permite insertar variables o expresiones entre llaves dentro de un texto.

```python
curso = "Big Data"
print("Hola, Python")
print(f"Curso: {curso}")
```

## 2. Variables y tipos de datos

Una variable es un nombre asociado a un valor. Usamos `=` para asignar ese valor y `type()` para consultar su tipo.

| Tipo | Qué representa | Ejemplo |
| --- | --- | --- |
| `int` | Número entero | `3` |
| `float` | Número con decimales | `1.5` |
| `str` | Texto | `"Big Data"` |
| `bool` | Valor lógico | `True` |

```python
sesiones = 3
horas_por_sesion = 1.5
es_practico = True

print(type(sesiones))
print(type(horas_por_sesion))
print(type(curso))
print(type(es_practico))
```

## 3. Operaciones y conversiones

Los operadores `+`, `-`, `*` y `/` permiten sumar, restar, multiplicar y dividir. Con `int` y `float`, Python admite operaciones mixtas. La división con `/` devuelve un `float` incluso cuando ambos operandos son enteros.

```python
print(3 + 2)
print(3 - 2)
print(3 * 2)
print(6 / 2)
print(3 + 1.5)
print(3 - 1.5)

horas_totales = sesiones * horas_por_sesion
print(f"Duración total: {horas_totales} h")
print(horas_totales / sesiones)
```

Con dos cadenas, `+` concatena texto. La expresión `"3" + 2` produce un `TypeError`: primero hay que decidir si queremos sumar números o unir texto. `int()` y `float()` convierten texto numérico y `str()` convierte un valor a texto.

```python
print("3" + "2")
print(int("3") + 2)
print(float("1.5") * sesiones)
print("Número de sesiones: " + str(sesiones))
```

## 4. Listas, tuplas y diccionarios

Las listas y las tuplas se consultan por posición. El primer elemento tiene índice `0`. El índice `-1` accede al último y un tramo como `[:2]` toma los dos primeros.

### Listas

Una lista se escribe con corchetes y permite modificar o añadir elementos.

```python
temas = ["Python", "PySpark", "DataFrames"]
print(temas[0])
print(temas[-1])
print(temas[:2])

temas[0] = "Primeros pasos con Python"
temas.append("Visualización")
print(temas)
```

### Tuplas

Una tupla agrupa valores en un orden fijo. Podemos leer sus posiciones o asignar sus elementos a varias variables en una sola línea.

```python
sesion = ("Python", 1.5)
print(sesion[0])

tema, duracion = sesion
print(f"{tema}: {duracion} h")
```

### Diccionarios

Un diccionario asocia claves con valores. Usamos `registro["tema"]` para consultar una clave y `registro["horas"] = 2.0` para actualizar un valor.

```python
registro = {"tema": "Python", "horas": 1.5}
print(registro["tema"])

registro["horas"] = 2.0
print(registro)
```

### Recorrer una colección con `for`

Un bucle `for` toma un elemento cada vez. En este ejemplo, cada elemento es un diccionario. La línea indentada se ejecuta una vez por sesión.

```python
programa = [
    registro,
    {"tema": "PySpark", "horas": 1.5},
    {"tema": "DataFrames", "horas": 2.0},
]

for sesion in programa:
    print(f"{sesion['tema']}: {sesion['horas']} h")
```

## 5. Primer DataFrame

Un DataFrame organiza datos en filas y columnas. Usaremos **pandas** para convertir la lista de diccionarios en una tabla. Cada diccionario aporta una fila y sus claves se convierten en nombres de columnas. Más adelante veremos cómo trabajar con DataFrames de Spark.

Si tu kernel muestra `ModuleNotFoundError` al importar pandas, ejecuta `%pip install pandas` en una celda y vuelve a probar la importación.

```python
import pandas as pd

tabla = pd.DataFrame(programa)
tabla
```

Seleccionamos una columna por su nombre. Con `.iloc[0]` accedemos a la primera fila por posición. Al multiplicar una columna numérica por 60, pandas aplica la operación a cada fila.

```python
print(tabla["tema"])
print(tabla.iloc[0])

tabla["minutos"] = tabla["horas"] * 60
print(f"Total: {tabla['minutos'].sum():.0f} minutos")
tabla
```

# Primeros pasos con PySpark

En esta práctica vamos a repartir una lista de datos, cambiar sus valores, filtrar y pedir un resultado. Antes de empezar, completa la [guía del entorno de PySpark](../01_environment_pyspark.md). Ejecuta las celdas en orden.

## 1. Iniciar Spark

La sesión es la que usamos para trabajar con Spark. De ella obtenemos el contexto, que aquí nos permite repartir una lista en varias partes. Con `local[2]`, Spark puede ejecutar hasta dos tareas a la vez en una sola máquina.

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.master("local[2]").appName("Primeros pasos con PySpark").getOrCreate()
sc = spark.sparkContext
print(f"Spark {spark.version} está listo.")
```

## 2. Crear un RDD con los pedidos del mercado

Cada número indica cuántas unidades hay en un pedido. `parallelize` convierte la lista en un RDD y la divide en dos partes, llamadas particiones. Aunque estamos en una sola máquina, Spark puede trabajar con esas partes por separado.

```python
pedidos = [2, 5, 3, 8, 1, 6]
pedidos_rdd = sc.parallelize(pedidos, 2)

print("Particiones:", pedidos_rdd.getNumPartitions())
```

## 3. Transformar los datos

Imagina que añadimos una unidad a cada pedido y queremos quedarnos con los pedidos de al menos 5 unidades. `map` aplica un cambio a cada valor y `filter` conserva los valores que cumplen una condición. Cada paso crea un RDD nuevo. Spark prepara estas instrucciones y las ejecuta cuando pedimos un resultado.

```python
pedidos_actualizados = pedidos_rdd.map(lambda unidades: unidades + 1)
pedidos_grandes = pedidos_actualizados.filter(lambda unidades: unidades >= 5)
```

## 4. Pedir resultados

`collect()` trae los valores al cuaderno y `count()` cuenta cuántos hay. Ambas son acciones. Aquí podemos usar `collect()` porque solo tenemos seis pedidos. Con datos grandes, usa `take(n)` para mirar unos pocos valores.

**Antes de ejecutar:** ¿qué pedidos deberían quedar y cuántos son?

```python
print("Pedidos seleccionados:", pedidos_grandes.collect())
print("Cantidad:", pedidos_grandes.count())
```

## 5. Probar la caché

Si vamos a consultar el mismo RDD varias veces, `cache()` le indica a Spark que guarde las partes que calcule para poder reutilizarlas. El primer conteo calcula los datos y llena la caché. En una lista tan pequeña, el tiempo de ejecución puede variar.

```python
pedidos_grandes.cache()

print("Primer conteo:", pedidos_grandes.count())
print("Segundo conteo:", pedidos_grandes.count())

pedidos_grandes.unpersist()
```

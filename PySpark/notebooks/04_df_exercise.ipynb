## Actividad 0 · Cargar el CSV

Ejecuta la celda de preparación para leer el [CSV incluido](../data/2024_pad_nai_mdbas.csv). Si usas Colab alojado por Google, sube el CSV al panel **Archivos**. En la VM, el cuaderno busca el archivo en la carpeta `data` del repositorio.

```python
from pathlib import Path

from pyspark.sql import SparkSession

archivo = "2024_pad_nai_mdbas.csv"
rutas = [Path("data") / archivo, Path("../data") / archivo, Path(archivo)]
ruta = next((opcion for opcion in rutas if opcion.is_file()), None)
if ruta is None:
    raise FileNotFoundError(f"No se encontró {archivo}. Revisa la carpeta data o sube el CSV a Colab.")

spark = SparkSession.builder.master("local[2]").appName("Nacimientos en Barcelona").getOrCreate()
sdf = spark.read.csv(str(ruta.resolve()), header=True, inferSchema=True, nullValue="..")
```

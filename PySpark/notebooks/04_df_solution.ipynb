# Ejercicios de DataFrames con datos de nacimientos

Explora los registros de nacimientos de Barcelona en 2024. La columna `Valor` recoge el recuento de nacimientos. El CSV usa `..` para algunos valores. Los leeremos como valores ausentes. Completa la [guía del entorno de PySpark](../01_environment_pyspark.md) si todavía no puedes iniciar Spark.

## Actividad 0 · Cargar el CSV

Ejecuta la celda de preparación para leer el [CSV incluido](../data/2024_pad_nai_mdbas.csv). Si usas Colab alojado por Google, sube el CSV al panel **Archivos**. En la VM, el cuaderno busca el archivo en la carpeta `data` del repositorio.

```python
from pyspark.sql import SparkSession

ruta = "../data/2024_pad_nai_mdbas.csv"

spark = SparkSession.builder.master("local[2]").appName("Nacimientos en Barcelona").getOrCreate()
sdf = spark.read.csv(ruta, header=True, inferSchema=True, nullValue="..")
```

### Nota sobre la Actividad 12

> **Observación Crítica:** Note que, aunque PySpark calcula la media (`mean`) y la desviación estándar (`stddev`) para `Codi_Districte`, estas métricas **no son estadísticamente significativas** ni útiles.
>
> Esto se debe a que `Codi_Districte` es un **código nominal** (una etiqueta categórica). Solo el **conteo (`count`)**, el **mínimo (`min`)** y el **máximo (`max`)** tienen sentido. Para analizar la distribución de categorías, siempre es mejor usar el método **`.groupBy().count()`**.
